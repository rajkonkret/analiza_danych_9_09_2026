```python
# lista - zachowuje kolejność
chaotyczna_lista = ['Tomek', 12, 45.6777, True, nowa_liczba_cal]
print(chaotyczna_lista)
```

```python
liczby = [43, 546, 324, 65, 12, 76, 3, 2345, 456, 788]  # 0,1,2,3,4
```

```python
liczby
```

```python
print(liczby[0])
```

```python
print(liczby[len(liczby) - 1])
```

```python
print(liczby[-1]) # ostatni element z kolekcji
```

```python
# slicing - fragment listy
print(chaotyczna_lista[1:3]) # indeksy 1 i 2, niewłącznie
```

```python
# od indeksu 2 do 8 włacznie
print(liczby[2:9])
```

```python
print(liczby[0:3])
```

```python
print(liczby[:3])
```

```python
print(liczby[2:]) # z ostatnim włacznie
```

```python
print(len(liczby))
```

```python
print(liczby[2:9])  # niewłacznie
```

```python
print(liczby[10])
```

```python
print(liczby[10:30])  # pusta lista
```

```python
chaotyczna_lista.append("lunch")
print(chaotyczna_lista)
```

```python
chaotyczna_lista.insert(2, "Radek")  # na indeksie 2
print(chaotyczna_lista)
```

```python
chaotyczna_lista[0] = "Przerwa"
print(chaotyczna_lista)
```

```python
chaotyczna_lista.remove(12)
print(chaotyczna_lista)
```

```python
# usunięcie po indeksie
print(chaotyczna_lista.pop(3))
```

```python
liczby.sort(reverse=True)
print(liczby)
```

```python
# posortowac chaotyczna lista
chaotyczna_lista.sort()
```

```python
# list comprehensions tansze i szybsze
mnoznik = [liczba * 2 for liczba in liczby]
print(mnoznik)

mnoznik_loop = [] # pusta lista
for liczba in liczby: # wykona 10 razy
    mnoznik_loop.append(liczba * 2)
print("Wyszlismy z pętli")   # pojawi się 1 raz
```

```python
print(type(mnoznik))
```

```python
print(mnoznik_loop)
```

 set - zbiór \
przechowuje unikalne wartości
nie zachowuje kolejności, nie ma indeksu

```python
lista = [23, 23, 45, 67, 89, 0, 23, 34]
lista.sort(reverse=True)
print(lista)
```

```python
# usunąc 23
lista.remove(23)
print(lista)
```

```python
zbior = set(lista)
print(zbior)
```

```python
pusty_zbior = set()  # tylko i wyłącznie za pomocą słowka
print(pusty_zbior)
print(type(pusty_zbior))
```
