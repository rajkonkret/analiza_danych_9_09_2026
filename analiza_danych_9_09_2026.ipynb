Klasa - przepis, szablon, stempel \
obiekt - zbudowany wg klasy, odcisk stempla, przysłowiowe ciasto

```python
class User:
    """Komentarz - dokumentacja"""

    def __init__(self, name, id):
        """Metoda inicjalizująca (konstruktor)"""
        self.name = name
        self.id = id

    def powitanie(self):
        print("Nazywam się:", self.name)

    def __str__(self):
        return f"{self.name} {self.id}"
```

```python
# pydoc
```

```python
# cz1 = User()
cz1 = User("Radek", 1)
```

```python
print(cz1)  # <__main__.User object at 0x7f9143686f90>, pododaniu __str__ Radek 1
```

```python
print(cz1.name)
```

```python
cz1.powitanie()
```

Przykład analizy

```python
sprzedaz = np.array([120, 135, 98, 150, 180, 110, 95, 210,170,140])
print(sprzedaz)
```

```python
# wypisac podstawowe miary
print("Średnia:", np.mean(sprzedaz))
```

```python
print("Mediana:", np.median(sprzedaz))
print("Minimum:", np.min(sprzedaz))
print("Maksimum:", np.max(sprzedaz))
print("Odchylenie:", np.std(sprzedaz))
print("Suma:", np.sum(sprzedaz))
```

```python
# duza sprzedaz - większa od średniej
duza_sprzedaz = sprzedaz[sprzedaz > np.mean(sprzedaz)]
print(duza_sprzedaz)
```

```python
# ile razy była duza sprzedaz
dni_duzej_sprzedazy = len(duza_sprzedaz)
print(dni_duzej_sprzedazy)
```

```python
wzrost_sprzedazy = sprzedaz * 1.10
print(wzrost_sprzedazy)
```

```python
# 3 sklepy, 4 miesiace

sprzedaz = np.array(
    [
        [120, 130, 140, 150],
        [100, 110, 115, 125],
        [200, 190,210, 220]
    ]
)

print(sprzedaz)
```

```python
# oblicz:
# suma dla kazdego sklepu
# suma dla kazdego miesiąca
# najlepszy sklep
```

```python
suma_sklep = sprzedaz[0].sum()
print(suma_sklep)
```

```python
print(np.sum(sprzedaz[0]))
```

```python
suma_sklepy_wiersze = sprzedaz.sum(axis=1) # per sklep
print("Suma dla każdego sklepu:", suma_sklepy_wiersze)
```

```python
suma_miesiacce_kolumna = sprzedaz.sum(axis=0)
print("Suma dla każdego miesiąca:", suma_miesiacce_kolumna)
```

```python
# najlpeszy sklep
print(np.max(suma_sklepy_wiersze))
```

```python
najlepszy_sklep = np.argmax(suma_sklepy_wiersze)
print("Najlepszy sklep:", najlepszy_sklep  + 1)
```
