```python
df['price'].fillna(df['price'].mean(), inplace=True)
```

```python
df.describe()
```

```python
df.isnull().sum()
```

----

```python
# wartości odstające
# IQR - kwartyle
```

```python
Q1 = df['score'].quantile(0.25)
Q3 = df['score'].quantile(0.75)
```

```python
IQR = Q3 - Q1
IQR
```

```python
outliers = df[df['score'] > Q3 + 1.5 * IQR]
outliers
```

```python
median_score = df['score'].median()
df.loc[df['score'] > Q3 + 1.5 * IQR, "score"] = median_score
df
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('orders.csv')
```

```python
df
```

```python
df = df.drop(columns=["total"]) # usunięcie kolumny
df
```

```python
# dodanie nowej kolumny, nowego parametru
df['total'] = df['price'] * df['quantity']
df
```

```python
# wyszukanie wiekrszy "Warszawa"
df.loc[df['city'] == "Warszawa"]
```

```python
# total > 4000
df.loc[df['total'] > 4000]
```

```python
df.loc[df['city'] == "Kraków", ["customer", 'city', 'price']]
```

```python
df.loc[df["customer"] == "Jan", ['customer', "price"]]
```

```python
df.loc[(df['category'] == "Laptop") & (df['paid'] == True),
       ['customer', 'category', 'total', 'paid']]
```

```python
# może prościej
df.loc[(df['category'] == "Laptop") & df['paid'],
       ['customer', 'category', 'total', 'paid']]
```

```python
# Warszawa i Gdańsk
df.loc[df['city'].isin(["Warszawa", "Gdańsk"])]
```

```python
# ~ not
df.loc[~df['city'].isin(["Warszawa", "Gdańsk"])]
```

```python
# isna()
df.loc[df['discount'].isna()]
```

```python
# notna()
df.loc[df['discount'].notna()]
```

```python
# zamien rabat NaN na 0
df.loc[df['discount'].isna(), "discount"] = 0
```

```python
df
```

```python
# W pole status wpisz zapłacone gdzie paid jest True
df.loc[df['paid'] == True, 'paid'] = 'zapłacone'
```

```python
df
```

```python
df.loc[df['paid'] == "zapłacone", 'paid'] = True
```

```python
df
```

```python
df.loc[df['paid'], 'status'] = "zapłacone"
```

```python
df
```

```python
# dla False -> niezapłacone
# df.loc[df["paid"] == False, 'status'] = 'niezapłacone'
df.loc[~df["paid"], 'status'] = 'niezapłacone' # wywoła błąd
```

```python
df
```

```python
# Dla zamówień powyżej 5000 wpisz w kolumnie segment wartość "VIP"
df.loc[df['total'] > 5000, "segment"] = "VIP"
df
```

```python
# dla mniejszych, równych 5000 wpisac "normal"
df.loc[df['total'] <= 5000, 'segment'] = 'normal'
```
