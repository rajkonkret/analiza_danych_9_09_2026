```python
import numpy as np
```

```python
input_stock_price_array = np.genfromtxt(
    'AAPL_stock_price_example.csv',
    delimiter=",",
    names=True, # pierwszy wiersz traktuja jako nazwy kolumn
    usecols=(1,) # wczytaj kolumne index 1
)
```

```python
print(input_stock_price_array)
```

```python
print(input_stock_price_array.size)
```

```python
# pięc pierwszych
print(input_stock_price_array[:5])
```

```python
# '<f8' eloat 8 bajtowy -> 64
input_stock_price_array.dtype
```

```python
# [::-1] - odwróćenie kolekcji
sorted_stock_price = np.sort(input_stock_price_array)[::-1]
print(sorted_stock_price)
```

```python
miasta = ["Kielce", "Poznań", 'Warszawa', 'Kraków']
```

```python
product_data = {
    "product a" : [13, 20, 0, 10],
    "product b": [10, 30, 17, 10],
    "product c" : [6, 9, 10, 0]
}
```

```python
# tworzymy DataFrame(tabela)
data = pd.DataFrame(product_data, index=miasta)
```

```python
print(data)
```

```python
data
```

```python
product_sales_sum = data.sum()
display(product_sales_sum)
```
