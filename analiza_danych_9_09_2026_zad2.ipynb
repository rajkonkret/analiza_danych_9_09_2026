```python
print(df.loc[2, "price"])
pd.isna(df.loc[2, "price"])
```

```python
df['price'].map(type).eq(str)
```

```python
df.head(8)
```

```python
df.isnull().sum()
```

```python
df['price'].fillna(df['price'].mean(), inplace=True)
```

```python
# df[col] = df[col].method(value)
# zalecana wersja
df['price'] = df['price'].fillna(df['price'].mean())
```

```python
df['price']
```

```python
df['score']
```

```python
df['score'] = df['score'].fillna(df['score'].median())
df['score']
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

----

```python
# wartości odstające
# IQR - kwartyle
```

```python
Q1 = df['score'].quantile(0.25)
Q3 = df['score'].quantile(0.75)
```

```python
IQR = Q3 - Q1
IQR
```

```python
outliers = df[df['score'] > Q3 + 1.5 * IQR]
outliers
```

```python
median_score = df['score'].median()
df.loc[df['score'] > Q3 + 1.5 * IQR, "score"] = median_score
df
```

```python
# wypisac posortowac wg order_date
df.sort_values(by="order_date")
```

```python
# score > 85 dla Poland
df[df['country'] == "Poland"]
```

```python
df['score'] > 85
```

```python
df[(df['country'] == "Poland") & (df['score']> 85)]
```

```python
df.query("category == 'Electronics' and price > 1000")
```

```python
# pogrupowac wg actegory order_id -> count
df.groupby('category').agg({'order_id': 'count'})
```

```python
# order_id, score -> mean
df.groupby('category').agg({'order_id': 'count', "score" :"mean"})
```

```python
# pivot table
df.pivot_table(values='price', index="country", aggfunc="mean")
```
